```python
models = ["llama", "mixtral"]
dataset = "EPIC"

for model in models:
    file_path = f"./results_label_change/label_change_{model}_{dataset}.txt"
    with open (file_path, "w") as f: 
        with redirect_stdout(f):
            prediction_change(dataset, datasets_vanilla, test_set, label, model, store_traits, lamp=False)

```
