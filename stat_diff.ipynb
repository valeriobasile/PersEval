```python
from perseval.evaluation import *
from perseval.data import *

# options for labels:
# Epic   -> ["irony"]
# BREXIT -> ["hs", "offensiveness", "aggressiveness", "stereotype"]
# DICES  -> ["Q2_harmful_content_overall"]
# MHS    -> ["hateful"]
# MD     -> ["offensiveness"]

label = "irony"
perspectivist_dataset = Epic(label)
perspectivist_dataset.get_splits(user_adaptation="train", extended=False, named=True)
```

```python
test_set = perspectivist_dataset.test_set
```

```python
store_traits = {}
for user, user_class in test_set.users.items():
    store_traits[user] = {}
    for k,v in user_class.traits.items():
        store_traits[user][k] = v[0]
```

```python
import csv

prediction_file = "/home/marem/VScProjects/PersEval/predictions_lamp_llama/edited_Epic_False.csv"

data = {}
with open(prediction_file, newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)

    for row in reader:
        key = (row["user_id"], row["text_id"])
        data[key] = {
            "gold": row["gold"],
            "pred": row["predictions"]
        }
        
# count_dict = {}

data
```

```python
for (user,text), value in data.items():
    for u,t in store_traits.items():
        if user == str(u): 
            value.update(t)

data
```

```python
from collections import defaultdict

list_traits = ["Gender", "Nationality", "Generation"]

result_pred = defaultdict(lambda:defaultdict(lambda: [0,0]))
result_gold = defaultdict(lambda:defaultdict(lambda: [0,0]))

for (user,text), value in data.items():
    pred = int(value["pred"])
    gold = int(value["gold"])
    for trait in list_traits: 
        value_trait = value[trait]
        if pred == 0:
            result_pred[text][value_trait][0] +=1
        elif pred == 1: 
            result_pred[text][value_trait][1] +=1

        if gold == 0:
            result_gold[text][value_trait][0] +=1
        elif gold == 1:
            result_gold[text][value_trait][1] +=1

result_gold
```
